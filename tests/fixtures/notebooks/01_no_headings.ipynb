Quick notes on pandas grouping. I keep forgetting the syntax.

```python
import pandas as pd
df = pd.DataFrame({'k': [1, 1, 2], 'v': [3, 4, 5]})
```

Group by a key, then aggregate with the mean of each group.

```python
df.groupby('k')['v'].mean()
```

Remember that groupby drops rows whose key is NaN by default.
